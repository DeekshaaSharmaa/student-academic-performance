```python
# copy() makes a separate table, so df (the raw data) is never changed
df_clean = df.copy()
```

```python
# drop_duplicates() removes rows that are exact copies of earlier rows
# reset_index() renumbers the rows 0, 1, 2... after rows are removed
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
print("Rows after removing duplicates:", len(df_clean))   # should be 200
```

```python
# strip() removes extra spaces, lower() makes everything lowercase
# map() then replaces each spelling with one standard value
df_clean["Gender"] = (
    df_clean["Gender"].str.strip().str.lower()
    .map({"male": "Male", "female": "Female", "f": "Female"})
)
print(df_clean["Gender"].value_counts())
```

```python
# Attendance above 100 is impossible, so we replace it with NaN (missing)
df_clean.loc[df_clean["Attendance_Percent"] > 100, "Attendance_Percent"] = np.nan
print(df_clean["Attendance_Percent"].max())
```

```python
cols = ["Attendance_Percent", "Study_Hours_Per_Week", "Mathematics", "Statistics"]
for col in cols:
    # median is the middle value, so it is not pulled by extreme values
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print(df_clean.isnull().sum())   # everything should now be 0
```

```python
import os
print("Shape:", df_clean.shape)
print("Duplicates:", df_clean.duplicated().sum())

os.makedirs("data/processed", exist_ok=True)       # creates the folder if missing
df_clean.to_csv("data/processed/student_performance_clean.csv", index=False)
print("Clean file saved!")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/processed/student_performance_clean.csv")
print(df.shape)       # should be (200, 10)
df.head()
```

```python
# a list of the five subject columns, so we can reuse it
subjects = ["Mathematics", "Python_Programming", "Computer_Fundamentals",
            "English", "Statistics"]

# agg() applies several functions at once; round(2) keeps 2 decimal places
summary = df[subjects].agg(["mean", "median", "min", "max", "std"]).round(2)
summary.T      # .T flips the table so each subject is one row
```

```python
subject_means = df[subjects].mean().round(2)
print("Highest average:", subject_means.idxmax(), "-", subject_means.max())
print("Lowest average: ", subject_means.idxmin(), "-", subject_means.min())
```

```python
df[["Attendance_Percent", "Study_Hours_Per_Week"]].agg(
    ["mean", "median", "min", "max", "std"]).round(2)
```

```python
# mean(axis=1) averages across the columns of each row (one student)
# This is a NEW calculated column, not part of the original dataset
df["Average_Marks"] = df[subjects].mean(axis=1).round(2)
df["Average_Marks"].describe().round(2)
```
